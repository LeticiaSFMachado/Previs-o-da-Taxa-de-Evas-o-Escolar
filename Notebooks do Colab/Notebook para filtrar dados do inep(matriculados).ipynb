```python
import pandas as pd
import os

# Pasta onde serão salvos os CSVs filtrados
pasta_saida = "/content/sample_data/matriculados"

os.makedirs(pasta_saida, exist_ok=True)

municipios_alvo = [
    'MARICÁ',
    'NITERÓI',
    'RIO DE JANEIRO',
    'SÃO GONÇALO'
]

colunas_necessarias = [
    'NU_ANO_CENSO',
    'NO_MUNICIPIO',
    'CO_ENTIDADE',
    'NO_ENTIDADE',
    'TP_DEPENDENCIA',
    'QT_MAT_MED'
]

def filtrar_matriculados(ano):

    if ano == 2025:
        arquivo = 'Tabela_Matricula_2025_V2.csv'
    else:
        arquivo = f'microdados_ed_basica_{ano}.csv'

    df = pd.read_csv(
        arquivo,
        sep=';',
        encoding='latin1',
        usecols=colunas_necessarias,
        low_memory=False
    )

    # Padroniza os nomes dos municípios
    df['NO_MUNICIPIO'] = (
        df['NO_MUNICIPIO']
        .astype(str)
        .str.strip()
        .str.upper()
    )

    # Aplicação dos filtros
    df_filtrado = df[
        (df['TP_DEPENDENCIA'] == 2) &
        (df['NO_MUNICIPIO'].isin(municipios_alvo)) &
        (df['QT_MAT_MED'] > 0)
    ].copy()

    # Renomeia as colunas
    df_filtrado = df_filtrado.rename(columns={
        'NU_ANO_CENSO': 'Ano do Censo',
        'NO_MUNICIPIO': 'Nome do Município',
        'CO_ENTIDADE': 'Código da Escola',
        'NO_ENTIDADE': 'Nome da Escola',
        'TP_DEPENDENCIA': 'Dependência Administrativa',
        'QT_MAT_MED': 'Quantidade de Matriculados Ensino Médio'
    })

    # Nome do arquivo de saída
    saida = f'{pasta_saida}/Matriculados{ano}ano.csv'

    df_filtrado.to_csv(
        saida,
        index=False,
        encoding='utf-8-sig'
    )

    print(f'Ano {ano}: {len(df_filtrado)} registros encontrados.')
    print(f'Arquivo salvo em: {saida}')

    return df_filtrado
```

# CONCAT

```python
import glob
import pandas as pd

# Pega somente os arquivos anuais de matriculados
pasta = "/content/sample_data/matriculados/Matriculados*ano.csv"

arquivos = sorted(glob.glob(pasta))

print(f"{len(arquivos)} arquivos encontrados:")
for arquivo in arquivos:
    print(arquivo)

if len(arquivos) > 0:

    df_final = pd.concat(
        [pd.read_csv(arquivo) for arquivo in arquivos],
        ignore_index=True
    )

    # Organiza a tabela
    df_final = df_final.sort_values(
        by=[
            'Ano do Censo',
            'Nome do Município',
            'Nome da Escola'
        ]
    ).reset_index(drop=True)

    print("\nConcatenação e ordenação concluídas!")
    print(f"Total de registros: {len(df_final)}")

else:
    print("Nenhum arquivo encontrado. Verifique o caminho.")
```

```python
arquivo_concatenado = "/content/sample_data/matriculados/matriculados_concatenados.csv"

df_final.to_csv(
    arquivo_concatenado,
    index=False,
    encoding="utf-8-sig"
)

print("Arquivo criado com sucesso!")
print(arquivo_concatenado)
```

```python
import pandas as pd

matriculados = pd.read_csv(
    "/content/drive/MyDrive/matriculados_concatenados.csv"
)

rendimentos = pd.read_csv(
    "/content/drive/MyDrive/rendimentos_concatenados.csv"
)

print("COLUNAS MATRICULADOS:")
print(matriculados.columns.tolist())

print("\nCOLUNAS RENDIMENTOS:")
print(rendimentos.columns.tolist())
```

```python
import unicodedata
import re

def normalizar_coluna(nome):
    # Remove acentos
    nome = unicodedata.normalize('NFKD', str(nome))
    nome = nome.encode('ascii', 'ignore').decode('ascii')

    # Remove espaços nas pontas
    nome = nome.strip()

    # Troca espaços por _
    nome = nome.replace(' ', '_')

    # Remove caracteres estranhos
    nome = re.sub(r'[^A-Za-z0-9_]', '', nome)

    return nome


# Padroniza todas as colunas
matriculados.columns = [
    normalizar_coluna(c) for c in matriculados.columns
]

rendimentos.columns = [
    normalizar_coluna(c) for c in rendimentos.columns
]

print("MATRICULADOS:")
print(matriculados.columns.tolist())

print("\nRENDIMENTOS:")
print(rendimentos.columns.tolist())
```

```python
matriculados['Codigo_da_Escola'] = pd.to_numeric(
    matriculados['Codigo_da_Escola'],
    errors='coerce'
).astype('Int64')

rendimentos['Codigo_Escola'] = pd.to_numeric(
    rendimentos['Codigo_Escola'],
    errors='coerce'
).astype('Int64')

matriculados['Ano_do_Censo'] = pd.to_numeric(
    matriculados['Ano_do_Censo'],
    errors='coerce'
).astype('Int64')

rendimentos['Ano'] = pd.to_numeric(
    rendimentos['Ano'],
    errors='coerce'
).astype('Int64')

print("Tipos padronizados com sucesso!")
```

```python
duplicados_matriculados = matriculados.duplicated(
    subset=['Codigo_da_Escola', 'Ano_do_Censo']
).sum()

duplicados_rendimentos = rendimentos.duplicated(
    subset=['Codigo_Escola', 'Ano']
).sum()

print("Duplicados em Matriculados:", duplicados_matriculados)
print("Duplicados em Rendimentos:", duplicados_rendimentos)
```

```python
df_join = pd.merge(
    matriculados,
    rendimentos,
    left_on=['Codigo_da_Escola', 'Ano_do_Censo'],
    right_on=['Codigo_Escola', 'Ano'],
    how='inner'
)

print("JOIN concluído!")
print("Matriculados:", len(matriculados))
print("Rendimentos:", len(rendimentos))
print("Resultado do JOIN:", len(df_join))

display(df_join.head())
```

```python
nao_encontrados = matriculados.merge(
    rendimentos,
    left_on=['Codigo_da_Escola', 'Ano_do_Censo'],
    right_on=['Codigo_Escola', 'Ano'],
    how='left',
    indicator=True
)

nao_encontrados = nao_encontrados[
    nao_encontrados['_merge'] == 'left_only'
]

display(
    nao_encontrados[
        [
            'Ano_do_Censo',
            'Nome_do_Municipio',
            'Codigo_da_Escola',
            'Nome_da_Escola'
        ]
    ]
)
```
