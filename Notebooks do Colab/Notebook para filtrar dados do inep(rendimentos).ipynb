#ANO 2025

```python
import pandas as pd

# 1. Carrega o arquivo inteiro sem cabeçalho
df_raw = pd.read_excel('tx_rend_escolas_2025.xlsx', sheet_name='ESCOLAS', header=None)

# 2. O Segredo: Isolar apenas as linhas do Excel que contêm os subtítulos úteis (índices 4, 5 e 6).
# O .ffill(axis=1) preenche as células mescladas para a direita para não perdermos nenhuma informação.
cabecalho = df_raw.iloc[4:7].ffill(axis=1).fillna('')

# 3. Cria nomes de colunas limpos juntando essas três linhas
nomes_colunas = []
for col in cabecalho.columns:
    partes = []
    for val in cabecalho[col]:
        texto = str(val).strip()
        # Junta os textos ignorando células vazias e repetições
        if texto and (not partes or partes[-1] != texto):
            partes.append(texto)
    # O resultado será algo como: "Taxa de Aprovação - Ensino Médio - Total"
    nomes_colunas.append(' - '.join(partes))

# 4. Aplica os nomes criados ao DataFrame e corta os dados (que começam na linha 8 / índice 7)
df = df_raw.iloc[7:].copy()
df.columns = nomes_colunas

# 5. Filtra pelos municípios desejados
col_mun = [c for c in df.columns if 'Nome do Município' in c][0]
municipios_alvo = ['Niterói', 'São Gonçalo', 'Maricá', 'Rio de Janeiro']
df_filtrado = df[df[col_mun].isin(municipios_alvo)].copy()

# 6. Seleciona as colunas de ID e as de Taxa do Ensino Médio (Aprovação, Reprovação, Abandono)
colunas_id = [c for c in df.columns if 'Nome do Município' in c or 'Nome da Escola' in c]

colunas_taxas = [c for c in df.columns if
                 'Ensino Médio' in c and
                 'Total' in c and
                 any(taxa in c for taxa in ['Aprovação', 'Reprovação', 'Abandono'])]

# 7. Tabela final
df_final = df_filtrado[colunas_id + colunas_taxas]
display(df_final.head())
```

```python
df.info()
```

```python
# 1. Puxamos as colunas incluindo o índice 5 (Código da Escola)
indices_alvo = [0, 4, 5, 6, 8, 21, 39, 57]
df_limpo = df.iloc[:, indices_alvo].copy()

# 2. Renomeamos as colunas adicionando o Codigo_Escola no lugar certo
df_limpo.columns = [
    'Ano',
    'Municipio',
    'Codigo_Escola', # <-- Nova coluna para o seu JOIN futuro!
    'Escola',
    'Dependencia',
    'Aprovacao_EM_Total',
    'Reprovacao_EM_Total',
    'Abandono_EM_Total'
]

# 3. Aplicamos os filtros (Cidades desejadas + Rede Estadual)
municipios_alvo = ['Niterói', 'São Gonçalo', 'Maricá', 'Rio de Janeiro']

df_estadual = df_limpo[
    (df_limpo['Municipio'].isin(municipios_alvo)) &
    (df_limpo['Dependencia'] == 'Estadual')
].copy()

# 4. Limpeza final dos dados: converte os textos/tracinhos para números e NaN
colunas_taxas = ['Aprovacao_EM_Total', 'Reprovacao_EM_Total', 'Abandono_EM_Total']

for col in colunas_taxas:
    df_estadual[col] = pd.to_numeric(df_estadual[col], errors='coerce')

# Remove escolas estaduais que não tenham nenhum dado preenchido de Ensino Médio
df_final = df_estadual.dropna(subset=colunas_taxas, how='all').copy()

# Exibe a tabela pronta!
display(df_final.head(10))
```

```python
df_final.to_csv('Rendimento2025ano.csv', index=False)
```

```python
# 1. Puxamos as colunas incluindo o índice 5 (Código da Escola)
indices_alvo = [0, 4, 5, 6, 8, 21, 39, 57]
df_limpo = df.iloc[:, indices_alvo].copy()

# 2. Renomeamos as colunas adicionando o Codigo_Escola no lugar certo
df_limpo.columns = [
    'Ano',
    'Municipio',
    'Codigo_Escola', # <-- Nova coluna para o seu JOIN futuro!
    'Escola',
    'Dependencia',
    'Aprovacao_EM_Total',
    'Reprovacao_EM_Total',
    'Abandono_EM_Total'
]

# 3. Aplicamos os filtros (Cidades desejadas + Rede Estadual)
municipios_alvo = ['Niterói', 'São Gonçalo', 'Maricá', 'Rio de Janeiro']

df_estadual = df_limpo[
    (df_limpo['Municipio'].isin(municipios_alvo)) &
    (df_limpo['Dependencia'] == 'Estadual')
].copy()

# 4. Limpeza final dos dados: converte os textos/tracinhos para números e NaN
colunas_taxas = ['Aprovacao_EM_Total', 'Reprovacao_EM_Total', 'Abandono_EM_Total']

for col in colunas_taxas:
    df_estadual[col] = pd.to_numeric(df_estadual[col], errors='coerce')

# Remove escolas estaduais que não tenham nenhum dado preenchido de Ensino Médio
df_final = df_estadual.dropna(subset=colunas_taxas, how='all').copy()

# Exibe a tabela pronta!
display(df_final)
```

```python
df_final.to_csv('Rendimento2019ano.csv', index=False)
```
