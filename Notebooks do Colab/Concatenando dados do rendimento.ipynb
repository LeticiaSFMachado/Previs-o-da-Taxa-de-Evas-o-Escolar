```python
import glob
import pandas as pd

pasta = "/content/sample_data/rendimentos/*.csv"

arquivos = sorted(glob.glob(pasta))

print(f"Arquivos encontrados e ordenados: {arquivos}")

if len(arquivos) > 0:
    df_final = pd.concat(
        [pd.read_csv(arquivo) for arquivo in arquivos], ignore_index=True
    )

    # Garante que as linhas internas também fiquem ordenadas pela primeira coluna (ano)
    primeira_coluna = df_final.columns[0]
    df_final = df_final.sort_values(by=primeira_coluna).reset_index(drop=True)

    print("Concatenação e ordenação por ano concluídas!")
else:
    print("Nenhum arquivo encontrado. Verifique o caminho.")
```
